````python
# 8 == 1000
# 1101 = 13

# 13 = 1101
# 13 / 2 = 6 remainder 1
# 6 / 2 = 3 remainder 0
# 3 / 2 = 1 remainder 1

# 17 / 2 = 8 remainder 1
# 8 / 2 = 4 remainder 0
# 4 / 2 = 2 remainder 0
# 2 / 2 = 1 remainder 0
# 1 / 2 = 1 remainder 1
````

````python
N = 13
quotient = N // 2 ; remainder = N % 2
print(f'quotient = {quotient}, remainder = {remainder}')
while quotient > 0:
    remainder = quotient % 2
    print(f'quotient = {quotient}, remainder = {remainder}')
    quotient = quotient // 2 
````

````python
result=[]
N = 13
while N > 0:
    remainder = N % 2
    N = N // 2
    result.append(remainder)
print(result[::-1])
````
